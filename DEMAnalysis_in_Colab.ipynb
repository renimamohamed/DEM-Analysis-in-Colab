```python
# Import necessary packages
!pip install rasterio
!pip install geopandas
!pip install plotly
import os
import matplotlib.pyplot as plt
from matplotlib.colors import LightSource # used for hillshade plot
# Use geopandas for vector data and rasterio for raster data
import geopandas as gpd
import rasterio as rio
# Plotting extent is used to plot raster & vector data together
from rasterio.plot import plotting_extent
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
# read DEM data
# source: https://www.earthdata.nasa.gov/sensors/srtm
with rio.open("/content/DEM.tif") as src:
    raster = src.read()
```

```python
# Plot the original raster
plt.figure(figsize=(15, 6))
# to plot rgb bands
plt.imshow(raster.transpose(1, 2, 0))
plt.title('Digital Elevation Model')
cbar = plt.colorbar()
```

```python
print("Minimum raster value: ", raster.min())
print("Maximum raster value: ", raster.max())
```

```python
# Set the figure size
plt.figure(figsize=(8, 6))  # Adjust width and height as needed

# Create a histogram for list1
plt.hist(raster.flatten(), bins=10, alpha=0.5, color='blue')

# Add labels and a legend
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.grid()
plt.title("DEM values")
```

```python
# Hillshade plot

# Create a light source object
light = LightSource(azdeg=315, altdeg=45)

# Calculate the hillshade values
hillshade = light.hillshade(raster[0], vert_exag=1, dx=1, dy=1)
```

```python
# Set the figure size
plt.figure(figsize=(12, 8))  # Adjust width and height as needed

# Create a plot for the hillshade
plt.imshow(hillshade, cmap='gray')
plt.colorbar(label='Hillshade Intensity')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Hillshade Plot')

plt.show()
```

```python
# Create a figure and axis
fig, ax = plt.subplots(figsize=(8, 8))

# Overlay the hillshade with the DEM values
cmap = plt.get_cmap('terrain')  # Choose a colormap for the DEM
dem_plot = ax.imshow(raster[0], cmap=cmap)
ax.imshow(hillshade, cmap='gray', alpha=0.7)  # Overlay the hillshade with transparency

# Add colorbars for DEM
dem_colorbar = plt.colorbar(dem_plot, ax=ax, label='Elevation (meters)')

ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('Hillshade Overlay with DEM')

plt.show()
```

```python
# Make a 3D plot of the DTM with matplot lib
from mpl_toolkits.mplot3d.axes3d import *
import matplotlib.pyplot as plt
from matplotlib import cm
```

```python
# get raster bounds

# Open the raster file
with rio.open('/content/DEM.tif') as src:
    # Get the bounds of the raster
    bounds = src.bounds
    width = src.width
    height = src.height

# Print the bounds
print("Left:", bounds.left)
print("Right:", bounds.right)
print("Top:", bounds.top)
print("Bottom:", bounds.bottom)
```

```python
xCoord = np.linspace(bounds.left, bounds.right, num=width)
yCoord = np.linspace(bounds.bottom, bounds.top, num=height)
```

```python
X, Y = np.meshgrid(xCoord, yCoord)
Z =  raster[0]
X = np.ma.masked_array(X)
Y = np.ma.masked_array(Y)
```

```python
fig = plt.figure(figsize=(15, 10))
ax = plt.axes(projection='3d')
pnt3d = ax.contour3D(X, Y, Z, 50,cmap='afmhot')
#cbar=plt.colorbar(pnt3d)
ax.set_xlabel('lat')
ax.set_ylabel('lon')
ax.set_zlabel('elevation')
ax.view_init(70, 40)
```

```python
# 3D interactive plot
# We will plot only a small part of the raster because this is memory consuming
import plotly.graph_objects as go
```

```python
sub_raster_width = 1000
fig = go.Figure(data=[go.Surface(x=X[0:sub_raster_width, 0:sub_raster_width], y=Y[0:sub_raster_width, 0:sub_raster_width], z=Z[0:sub_raster_width, 0:sub_raster_width])])

fig.update_layout(title='DEM')

fig.show()
```
